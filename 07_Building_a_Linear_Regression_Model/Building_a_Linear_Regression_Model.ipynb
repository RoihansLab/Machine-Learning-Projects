## **Diabetes**

## **1. Generate datasets**

```python
from sklearn import datasets
```

### **load_datasets**



```python
diabetes = datasets.load_diabetes()
```

### **Examine the datasets**

```python
diabetes
```

```python
print(diabetes.DESCR)
```

```python
print(diabetes.feature_names)
```

```python
print(diabetes.target)
```

```python
import pandas as pd

df = pd.DataFrame(diabetes.data, columns=diabetes.feature_names)
df['target'] = diabetes.target
df
```

```python
X = diabetes.data
y = diabetes.target
```

```python
X.shape, y.shape
```

## **2. Split Datasets**

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

### **Examine Data train and test**

```python
X_train.shape, y_train.shape
```

```python
X_test.shape, y_test.shape
```

## **3. Modeling with Linear Regression**

### **Train the model**

```python
from sklearn import linear_model
from sklearn.metrics import mean_squared_error, r2_score, accuracy_score

LR = linear_model.LinearRegression()
LR.fit(X_train, y_train)
```

### **predict and evaluate**

```python
y_pred = LR.predict(X_test)

print("coefficien", LR.coef_)
print("intercept", LR.intercept_)
print(f'MSE: {mean_squared_error(y_test, y_pred):.2f}')
print(f'R2: {r2_score(y_test, y_pred):.2f}')
```

```python
y_pred, y_test
```

## **4. Visualized**

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.scatterplot(x=y_test, y=y_pred, alpha=0.5)
plt.xlabel('Actual')
plt.ylabel('Predicted')
plt.title('Actual vs. Predicted')
```

#### *LINEAR REGGRESSION TEORY*

**intercept** ($\beta_0$ / Titik Awal):
Ini ibarat "Tarif Buka Pintu". intercept adalah nilai tebakan paling dasar atau nilai prediksi target ($y$) ketika semua fitur data ($x$) nilainya benar-benar $0$.

**coef** ($\beta_1, \beta_2 \dots$ / Bobot Fitur):Ini ibarat "Tarif per Kilometer". coef (koefisien) menunjukkan seberapa kuat impact atau pengaruh dari setiap fitur ($x$) terhadap hasil prediksi ($y$).Kalau nilainya positif, setiap kenaikan angka pada fitur tersebut akan menaikkan hasil prediksi.Kalau nilainya negatif, setiap kenaikan angka pada fitur tersebut akan menurunkan hasil prediksi.


Dalam model Linear Regression (secara matematis ditulis sebagai $y = \beta_0 + \beta_1 x_1 + \dots + \beta_n x_n$)

exmpl:
pred/medv = -1.12*(crim) + 5.14*(zn) + 3.18*(indus) + .... + 39

***this is how it works to make a prediction***
