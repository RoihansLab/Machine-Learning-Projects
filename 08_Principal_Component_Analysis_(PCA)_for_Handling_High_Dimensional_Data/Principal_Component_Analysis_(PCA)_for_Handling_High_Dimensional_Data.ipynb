# **Principal Component Analysis(PCA) for Handling High Dimensional Data**

# **1. Iris Dataset**

## 1.1. Load Dataset

```python
from sklearn import datasets
```

## 1.2. Load Datasets

```python
iris = datasets.load_iris()
```

```python
iris
```

## 1.3. input features

```python
print(iris.feature_names)
```

## 1.4. output fetures

```python
print(iris.target_names)
```

## 1.5. Assigning Input(X) and Output (y) variables

```python
X = iris.data
y = iris.target
```

## 1.6. Examine the data dimension

```python
X.shape, y.shape
```

# **2. PCA Analysis**

## 2.1. Load Library

```python
from sklearn.preprocessing import scale
from sklearn import decomposition
import pandas as pd
```

## 2.2. Data Scaling

```python
X = scale(X)
```

## 2.3. Perform PCA analysis




we define the number of PC to use as 3

```python
pca = decomposition.PCA(n_components=3, random_state=42)
pca.fit(X)
```

## 2.4. Compute and Retrive the score values

```python
scores = pca.transform(X)
score_df = pd.DataFrame(scores, columns=['PC1', 'PC2', 'PC3'])
score_df
```

```python
y_label = []
for i in y:
  if i == 0:
      y_label.append('sentosa')
  elif i == 1:
      y_label.append('versicolor')
  else:
      y_label.append('Virginica')

species = pd.DataFrame(y_label, columns=['Species'])
```

```python
df_scores = pd.concat([score_df, species], axis=1)
df_scores
```

## 2.5. Retrive the loadings values

```python
loadings = pca.components_.T
df_loadings = pd.DataFrame(loadings, columns=['PC1', 'PC2', 'PC3'], index=iris.feature_names)
df_loadings
```

## 2.6. **Explained variance** for each **PC**

```python
explained_variance = pca.explained_variance_ratio_
explained_variance
```

# **3. Scree Plot**

## 3.1 Import Library

```python
import numpy as np
import plotly.express as px
```

## 3.2.Preparing the explained variance data

### 3.2.1. preparing the explained variance data

```python
explained_variance
```

```python
explained_variance = np.insert(explained_variance, 0, 0)
```

### 3.2.2 preparing the cumulative variance data

```python
cumulative_variance = np.cumsum(np.round(explained_variance, decimals=3))
```

### 3.2.3. Combining the DataFrame

```python
pc_df = pd.DataFrame(['', 'PC1', 'PC2', 'PC3'], columns=['PC'])
explained_variance_df = pd.DataFrame(explained_variance, columns=['Explained Variance'])
cumulative_variance_df = pd.DataFrame(cumulative_variance, columns=['Cumulative Variance'])
```

```python
df_explained_variance = pd.concat([pc_df, explained_variance_df , cumulative_variance_df], axis=1)
df_explained_variance
```

### 3.2.4. Making the scree plot

#### 3.2.4.1. Explained Variance

```python
#https://plotly.com/python/bar-charts/

fig = px.bar(df_explained_variance,
             x='PC', y='Explained Variance',
             text='Explained Variance',
             width=800)

fig.update_traces(texttemplate='%{text:.3f}', textposition='outside')
fig.show()
```
