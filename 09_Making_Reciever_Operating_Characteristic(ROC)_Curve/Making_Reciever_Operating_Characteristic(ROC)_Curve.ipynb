#**Machine Learning in python : Making Reciever Characteristic (ROC) Curve**

### What is ROC curve?

The ROC curve summarizes the prediction performance of a classification model at all classification thresholds. Particularly, the ROC curve plots the False Positive Rate (FPR) on the X-axis and the True Positive Rate (TPR) on the Y-axis.

$$TPR(Sensitivity) = \frac{TP}{TP+FN}$$

$$FPR(1 - Specificity) = \frac{FP}{TN+FP}$$

## **1. Generate Datasets**

```python
from sklearn.datasets import make_classification
import numpy as np
```

```python
X, y = make_classification(n_samples=2000, n_classes=2, n_features=10, random_state=0)
```

### add noisy features to make problem more difficult

```python
random_state = np.random.RandomState(0)
n_samples, n_features = X.shape
X = np.c_[X, random_state.randn(n_samples, 200 * n_features)]
```

## **2. Data Splitting**

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)
```

## **3. Build Classification Model**

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.naive_bayes import GaussianNB
```

Random Forerst

```python
rf = RandomForestClassifier(max_features= 5, n_estimators=100)
rf.fit(X_train, y_train)
```

Naive Bayes

```python
GNB = GaussianNB()
GNB.fit(X_train, y_train)
```

### **Prediction Propabilities**

```python
r_probs = [0 for _ in range (len(y_test))]
rf_probs = rf.predict_proba(X_test)
gnb_probs = GNB.predict_proba(X_test)
```

```python
rf_probs = rf_probs[:, 1]
gnb_probs = gnb_probs[:, 1]
```

```python
from sklearn.metrics import roc_auc_score
```

```python
r_auc = roc_auc_score(y_test, r_probs)
rf_auc = roc_auc_score(y_test, rf_probs)
gnb_auc = roc_auc_score(y_test, gnb_probs)
```

### **Print AUROC scores**

```python
print('random (chance)prediction (AUROC): %.3f' % (r_auc))
print('Random Forest (AUROC):  %.3f' %  (rf_auc))
print('Naive Bayes (AUROC):  %.3f' %  (gnb_auc))
```
