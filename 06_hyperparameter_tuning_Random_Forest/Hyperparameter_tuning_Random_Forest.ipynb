## **3. Building a Simple Machine Learning Model with Random Forest**

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

rf = RandomForestClassifier(n_estimators=100, max_features=100)
```

```python
rf.fit(X_train, y_train)
```

```python
rf.score(X_test, y_test)
```

```python
y_pred = rf.predict(X_test)
```

```python
accuracy_score(y_pred, y_test)
```

```python
y_pred, y_test
```

## **4. Hyperparametet Tuning**

now i will be performing of the tuning of hyperparamaters of RandomForestClassifier model. The hyperparameters that we will tune includes max_features and n_estimators

```python
from sklearn.model_selection import GridSearchCV
import numpy as np

max_features_range = np.arange(1,6,1)
n_estimators_range = np.arange(10,210,10)
param_grid = dict(max_features=max_features_range, n_estimators=n_estimators_range)

rf = RandomForestClassifier()

grid = GridSearchCV(estimator=rf, param_grid=param_grid, cv=5)
```

```python
grid.fit(X_train, y_train)
```

```python
print("The best parameters are %s with a score of %0.2f"
% (grid.best_params_, grid.best_score_))
```

## **5. Data Frame of Grid Search parameters and their Accuracy Scores**

```python
import pandas as pd

grid_results = pd.concat([pd.DataFrame(grid.cv_results_["params"]), pd.DataFrame(grid.cv_results_["mean_test_score"], columns=["Accuracy"])], axis=1)

grid_results
```

## **6. Preparing data for making contour plots**

prior to making contour plots, we will have to reshape the data into compatible that will be recognized by th contour plot functions.

I'll be using pandas groupby() function to segment the data into groups based on 2 parameters: max_features and n_enstimators

```python
grid_contour = grid_results.groupby(['max_features', 'n_estimators']).mean()
grid_contour
```

### **Pivoting the data**

Data is reshape by pivoting the data into an m matrix where rows and columns correspond to the max_features and n_estimators, respectively

```python
grid_reset = grid_contour.reset_index()
grid_reset.columns = ['max_features', 'n_estimators', 'Accuracy']
grid_pivot = grid_reset.pivot(index='max_features', columns='n_estimators', values='Accuracy')
grid_pivot
```

I assign the pivoted data into the perspective x, y and z

```python
x = grid_pivot.columns.values
y = grid_pivot.index.values
z = grid_pivot.values
```

```python
x, y, z
```

### **2D Contour Plot**

visualizing the landscape of the 2 hyperparameters that we are tuning and their influence on the accuracy

```python
import plotly.graph_objects as go

# Pakai go.Layout (L besar) dan dict() biar jauh lebih simpel!
layout = go.Layout(
    xaxis=dict(title='max_features'),
    yaxis=dict(title='n_estimators')
)

# Pakai go.Figure (F besar)
fig = go.Figure(data=[go.Contour(z=z, x=x, y=y)], layout=layout)

fig.update_layout(title='Hyperparameter tuning', autosize=False, width=500, height=500, margin=dict(l=65, r=50, b=65, t=90))

fig.show()
```

### **3D Surface Plot**

```python
import plotly.graph_objects as go

fig = go.Figure(data=[go.Surface(z=z, y=y, x=x)], layout=layout)

fig.update_layout(title='Hyperparameter tuning',
                  scene=dict(
                      xaxis_title='max_features',
                      yaxis_title='n_estimators',
                      zaxis_title='Accuracy'
                  ),
                  autosize=False,
                  width=500,
                  height=500,
                  margin=dict(l=65, r=50, b=65, t=90))

fig.show()
```
